```python
f=open("mbox-short.txt")
x=f.readlines()
s=0
c=0
for i in x:
    if "X-DSPAM-Confidence:" in i:
        k=i.split(" ")
        s=s+float(k[1])
        c=c+1
print(f"average value is {s/c}")
f.close()
```

```python
f=open("mbox-short.txt")
x=f.readlines()
c=0
for i in x:
    if "From " in i:
        k=i.split(" ")
        print(k[1])
        c=c+1
print(f"count of email is {c}")
```

```python
f=open("mbox-short.txt")
x=f.readlines()
l=[]
for i in x:
    if i.startswith("From "):
        k=i.split(" ")
        l.append(k[1])
c=0
d={}
for i in l:
    d[i]=l.count(i)
    c=c+1
print(d)
print(max(d.values()))
```
