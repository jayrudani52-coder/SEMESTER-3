```python
class wordplay:
    def __init__(self,l):
        self.l=l
        print(self.l)
    def words_with_length(self,l1):
        x=[]
        for i in self.l:
            if len(i)==l1:
                x.append(i)
        print(x)
    def starts_with(self,u1):
        x=[]
        for i in self.l:
            if i.startswith(u1):
                x.append(i)
        print(x)
    def ends_with(self,u2):
        x=[]
        for i in self.l:
            if i.endswith(u2):
                x.append(i)
        print(x)
    def palindromes(self):
        x=[]
        for i in self.l:
            if i[0]==i[len(i)-1]:
                x.append(i)
        print(x)
    def only(self,u4):
        x=set(u4)
        for i in self.l:
            z=set(i)
            if x.intersection(z)==x:
                print(i)
    def avoids(self,u5):
        x=set(u5)
        for i in self.l:
            z=set(i)
            if x.intersection(z)==set():
                print(i)
s=wordplay(['apple','banana','find','dictionary','set','tuple','list','malyalam','nayan','grind','apricot'])
s.words_with_length(5)
s.starts_with('a')
s.ends_with('d')
s.palindromes()
s.only('bna')
s.avoids('amkd')
```
